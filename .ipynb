```python
import os

def get_all_file_paths(root_folder="."):
    all_paths = []
    for root, dirs, files in os.walk(root_folder):
        # Skip .git and .venv folders completely
        dirs[:] = [d for d in dirs if d not in {".git", ".venv"}]

        for file in files:
            full_path = os.path.join(root, file)
            all_paths.append(os.path.abspath(full_path))
    return all_paths


if __name__ == "__main__":
    folder = "."  # current directory
    abs_folder = os.path.abspath(folder)
    print(f"🔍 Scanning folder: {abs_folder}\n")

    if not os.path.exists(folder):
        print("❌ Folder not found!")
    else:
        paths = get_all_file_paths(folder)
        print(f"📁 Total files found (excluding .git and .venv): {len(paths)}\n")
        for path in paths:
            print(path)
```
