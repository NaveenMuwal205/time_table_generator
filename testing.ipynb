```python
# cell 1
import re

def split_subject(text):
    """
    Splits subject name and subject code from input string.
    Works for flexible code formats like BTAI-702, MCA101, CSE-2021 etc.
    """
    # Regex to capture: name part + code part
    match = re.match(r"^(.*?)([A-Za-z0-9-]+)$", text.strip())

    if match:
        subject_name = match.group(1).strip()
        subject_code = match.group(2).strip()
        return subject_name, subject_code
    else:
        return text, None  # fallback if no code found

# Example usage
examples = [
    "Robotics and AI BTAI-702",
    "Computer Networks MCA101",
    "Data Science BCA-202",
    "Machine Learning CSE-9999"
]

for e in examples:
    name, code = split_subject(e)
    print(f"Input: {e}\n  Name: {name}\n  Code: {code}\n")
```
